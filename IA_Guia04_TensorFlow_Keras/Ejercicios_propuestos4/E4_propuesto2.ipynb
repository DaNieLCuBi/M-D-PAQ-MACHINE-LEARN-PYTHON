```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir el rango de valores para z (eje X)
z = np.linspace(-10, 10, 100)

# 2. Definir las funciones de activación
def sigmoide(z):
    return 1 / (1 + np.exp(-z))

def relu(z):
    return np.maximum(0, z)

def tanh(z):
    return np.tanh(z)

def leaky_relu(z, alpha=0.1):
    return np.where(z > 0, z, z * alpha)

# 5. Función Lineal (identidad: y = z)
def lineal(z):
    return z

# 3. Calcular los valores de salida para cada función
y_sigmoide = sigmoide(z)
y_relu = relu(z)
y_tanh = tanh(z)
y_leaky = leaky_relu(z)
y_lineal = lineal(z)

# 4. Crear la figura y las subgráfi-cas (grid de 2x3 o una fila grande)
plt.figure(figsize=(15, 10))

# Gráfica 1: Sigmoide
plt.subplot(2, 3, 1)
plt.plot(z, y_sigmoide, color='blue')
plt.title('Sigmoide')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

# Gráfica 2: ReLU
plt.subplot(2, 3, 2)
plt.plot(z, y_relu, color='red')
plt.title('ReLU')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

# Gráfica 3: Tanh
plt.subplot(2, 3, 3)
plt.plot(z, y_tanh, color='green')
plt.title('Tanh')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

# Gráfica 4: Leaky ReLU
plt.subplot(2, 3, 4)
plt.plot(z, y_leaky, color='orange')
plt.title('Leaky ReLU')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

# Gráfica 5: Lineal (Nueva)
plt.subplot(2, 3, 5)
plt.plot(z, y_lineal, color='purple')
plt.title('Lineal (y = z)')
plt.grid(True)
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

# Ajustar diseño y mostrar
plt.tight_layout()
plt.show()
```
