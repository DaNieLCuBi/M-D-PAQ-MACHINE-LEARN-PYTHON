```python
import numpy as np

# 1. Definir los pesos modificados: [horas_estudio, asistencia, tareas_entregadas]
# Damos un valor muy alto a la asistencia (5.0) para que sea el factor dominante.
w = np.array([0.2, 5.0, 0.1])
b = -3.0  # Sesgo o umbral de activación

# 2. Función de activación sigmoide
def sigmoide(z):
    return 1 / (1 + np.exp(-z))

# 3. Datos de los tres estudiantes
# Estudiante 1: Estudió poco, asistió casi siempre (0.9), pocas tareas.
x1 = np.array([1.0, 0.9, 1.0]) 

# Estudiante 2: Estudió muchísimo (9.0), pero asistencia pésima (0.1), varias tareas.
x2 = np.array([9.0, 0.1, 5.0]) 

# Estudiante 3: Rendimiento medio, asistencia intermedia (0.5).
x3 = np.array([4.0, 0.5, 3.0]) 

# 4. Cálculos y resultados para cada estudiante
salida1 = sigmoide(np.dot(x1, w) + b)
salida2 = sigmoide(np.dot(x2, w) + b)
salida3 = sigmoide(np.dot(x3, w) + b)

print(f"Estudiante 1 -> Probabilidad: {round(salida1, 3)} | Decisión: {'Aprueba' if salida1 >= 0.5 else 'Desaprueba'}")
print(f"Estudiante 2 -> Probabilidad: {round(salida2, 3)} | Decisión: {'Aprueba' if salida2 >= 0.5 else 'Desaprueba'}")
print(f"Estudiante 3 -> Probabilidad: {round(salida3, 3)} | Decisión: {'Aprueba' if salida3 >= 0.5 else 'Desaprueba'}")
```
