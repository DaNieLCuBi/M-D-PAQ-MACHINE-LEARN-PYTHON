```python
# 04_primera_red.py - La red más simple: aprender la regla y = 3x - 2
import numpy as np
import keras
from keras import layers

keras.utils.set_random_seed(42)  # resultados repetibles

# 1. Datos de entrenamiento: cambiamos la fórmula a y = 3x - 2
X = np.arange(-5, 6, dtype="float32")  # -5, -4, ..., 5
y = 3 * X - 2
print("x:", X)
print("y:", y)

# 2. Modelo: 1 entrada -> 1 neurona (sin activación = salida lineal)
modelo = keras.Sequential([
    keras.Input(shape=(1,)),
    layers.Dense(1),
])
modelo.compile(optimizer=keras.optimizers.SGD(learning_rate=0.01), loss="mse")

# 3. Entrenamiento: 500 épocas (vueltas completas sobre los datos)
historial = modelo.fit(X, y, epochs=500, verbose=0)
for epoca in [1, 100, 250, 500]:
    print(f"Época {epoca:3}: pérdida (MSE) = {historial.history['loss'][epoca - 1]:.4f}")

# 4. ¿Qué aprendió?
w, b = modelo.layers[0].get_weights()
print(f"Peso aprendido: {w[0][0]:.3f} (ideal 3) | Sesgo aprendido: {b[0]:.3f} (ideal -2)")

# Probando con un valor nuevo, por ejemplo x = 10 (ideal: 3(10) - 2 = 28)
prediccion = modelovale = modelo.predict(np.array([[10.0]]), verbose=0)
print(f"Predicción para x = 10: {modelovale[0][0]:.2f} (valor real 28)")
```
