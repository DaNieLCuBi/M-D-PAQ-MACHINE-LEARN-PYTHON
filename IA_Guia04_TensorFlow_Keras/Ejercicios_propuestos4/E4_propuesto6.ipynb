```python
# 06_cargar_modelo.py / Celda para Jupyter Notebook (E4_propuesto6.ipynb)
import numpy as np
import keras
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

# 1. Cargar el modelo usando la ruta relativa hacia atrás ("../") 
# para ubicarlo en la carpeta superior IA_Guia04_TensorFlow_Keras
modelo = keras.models.load_model("../modelo_iris.keras")
print("Modelo cargado exitosamente desde la carpeta superior.")

# 2. Reconstruir el escalador utilizando los datos originales de Iris 
# (Es obligatorio escalar la nueva flor exactamente igual que en el entrenamiento)
iris = load_iris()
X = iris.data
escalador = StandardScaler()
escalador.fit(X)

# 3. Definir y escalar la nueva flor: [6.7, 3.0, 5.2, 2.3]
flor_nueva = np.array([[6.7, 3.0, 5.2, 2.3]])
flor_escalada = escalador.transform(flor_nueva)

# 4. Hacer la predicción con el modelo
probabilidades = modelo.predict(flor_escalada, verbose=0)[0]
especie_predicha = iris.target_names[np.argmax(probabilidades)]

# 5. Mostrar los resultados
print("\n--- RESULTADO DE LA PREDICCIÓN ---")
print("Probabilidades por especie:")
for especie, prob in zip(iris.target_names, probabilidades):
    print(f" - {especie}: {prob * 100:.2f}%")

print(f"\nLa especie predicha para la flor [6.7, 3.0, 5.2, 2.3] es: **{especie_predicha}**")
```
