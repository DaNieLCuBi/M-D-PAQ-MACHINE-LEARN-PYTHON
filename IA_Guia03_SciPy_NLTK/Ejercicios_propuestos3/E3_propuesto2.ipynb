```python
import numpy as np
from scipy import optimize

# 1. Definir la función f(x) = x² - 6x + 13
def f(x):
    return x**2 - 6*x + 13

# 2. Encontrar el mínimo utilizando minimize_scalar
resultado = optimize.minimize_scalar(f)

# 3. Mostrar los resultados de SciPy
x_min = resultado.x
y_min = resultado.fun

print(" RESULTADO CON SCIPY ")
print(f"Mínimo encontrado en x = {x_min:.4f}")
print(f"Valor mínimo de f(x)  = {y_min:.4f}\n")

# 4. Comprobación matemática a mano
print(" COMPROBACIÓN MANUAL (DERIVADA) ")
print("1. Función original:   f(x) = x² - 6x + 13")
print("2. Primera derivada:   f'(x) = 2x - 6")
print("3. Igualar a cero:     2x - 6 = 0  =>  2x = 6  =>  x = 3")
print("4. Segunda derivada:   f''(x) = 2  (al ser > 0, se confirma que es un MÍNIMO)")
print(f"5. Evaluar en f(3):    f(3) = (3)² - 6(3) + 13 = 9 - 18 + 13 = {f(3)}")
```
