```python
import nltk
from nltk.stem import SnowballStemmer
from nltk.tokenize import word_tokenize
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy

# 1. Definir el conjunto de datos ampliado (Opiniones originales + 10 nuevas)
datos_entrenamiento = [
    # Opiniones iniciales (Paso 13)
    ("Excelente clase, aprendí mucho", "positivo"),
    ("El profesor explica muy bien y claro", "positivo"),
    ("Buena explicación y gran contenido", "positivo"),
    ("Pésima clase, no me gustó para nada", "negativo"),
    ("Muy confuso y nada claro el tema", "negativo"),
    ("No entendí nada de la explicación", "negativo"),
    
    # 5 Nuevas opiniones positivas
    ("La clase estuvo fantástica e interesante", "positivo"),
    ("Gran metodología y excelentes ejercicios", "positivo"),
    ("Disfruté mucho la lección de hoy", "positivo"),
    ("El tema es fascinante y muy bien explicado", "positivo"),
    ("Me encantó el curso, sumamente útil", "positivo"),
    
    # 5 Nuevas opiniones negativas
    ("La clase fue aburrida y monótona", "negativo"),
    ("Perdí el tiempo, estuvo horrible", "negativo"),
    ("No me agradó la presentación del tema", "negativo"),
    ("Poca claridad y contenido deficiente", "negativo"),
    ("La sesión fue muy aburrida y pesada", "negativo")
]

# 2. Inicializar el stemmer en español
stemmer = SnowballStemmer("spanish")

# 3. Función de extracción de características APLICANDO STEMMING
def extraer_caracteristicas(texto):
    tokens = word_tokenize(texto.lower(), language="spanish")
    # Convertir cada palabra a su raíz (stemming)
    raices = [stemmer.stem(palabra) for palabra in tokens if palabra.isalpha()]
    return {raiz: True for raiz in raices}

# 4. Procesar el conjunto de datos de entrenamiento
datos_procesados = [
    (extraer_caracteristicas(texto), etiqueta) 
    for (texto, etiqueta) in datos_entrenamiento
]

# 5. Entrenar el clasificador Naive Bayes
clasificador = NaiveBayesClassifier.train(datos_procesados)

# 6. Probar con la frase objetivo "La clase fue aburrida"
frase_prueba = "La clase fue aburrida"
caracteristicas_prueba = extraer_caracteristicas(frase_prueba)
resultado = clasificador.classify(caracteristicas_prueba)

# 7. Mostrar resultados
print("EVALUACIÓN DEL CLASIFICADOR")
print(f"Frase a evaluar: '{frase_prueba}'")
print(f"Características (Raíces extraídas): {list(caracteristicas_prueba.keys())}")
print(f"Resultado de la clasificación: {resultado.upper()}\n")

# Mostrar las características más informativas
print("CARACTERÍSTICAS MÁS INFORMATIVAS")
clasificador.show_most_informative_features(5)
```
