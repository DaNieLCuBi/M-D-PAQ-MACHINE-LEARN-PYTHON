```python
import numpy as np
from scipy import stats

# 1. Registrar las notas de dos cursos (al menos 8 notas por curso)
curso_matematica = [14, 16, 12, 15, 18, 11, 13, 17, 15, 16]
curso_programacion = [8, 10, 11, 9, 12, 10, 13, 7, 11, 10]

# 2. Aplicar la prueba t de Student para muestras independientes
t_stat, p_value = stats.ttest_ind(curso_matematica, curso_programacion)

# 3. Mostrar promedios y resultados estadísticos
print(" RESULTADOS DEL ANÁLISIS ")
print(f"Promedio Matemáticas:  {np.mean(curso_matematica):.2f}")
print(f"Promedio Programación: {np.mean(curso_programacion):.2f}")
print(f"Estadístico t:        {t_stat:.4f}")
print(f"Valor p (p-value):    {p_value:.6f}\n")

# 4. Interpretación del resultado (Nivel de significancia alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print("Respuesta: SÍ hay una diferencia estadísticamente significativa entre las notas de ambos cursos.")
else:
    print("Respuesta: NO hay una diferencia estadísticamente significativa entre las notas de ambos cursos.")
```
