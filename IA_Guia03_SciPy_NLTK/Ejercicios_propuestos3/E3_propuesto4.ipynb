```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist
# Asegurar la descarga de recursos necesarios de NLTK
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)
# 1. Párrafo de una noticia en español
noticia = (
    "La inteligencia artificial continúa transformando el sector tecnológico a nivel mundial. "
    "Múltiples empresas de tecnología e innovación están invirtiendo recursos significativos "
    "para desarrollar nuevos modelos de inteligencia artificial que optimicen procesos, "
    "mejoren la productividad e impulsen la innovación digital en diversos sectores de la industria.")
# 2. Tokenizar y limpiar (convertir a minúsculas y filtrar solo palabras)
tokens = word_tokenize(noticia.lower(), language="spanish")
palabras = [t for t in tokens if t.isalpha()]
# 3. Eliminar stopwords (palabras vacías)
vacias = set(stopwords.words("spanish"))
palabras_limpias = [p for p in palabras if p not in vacias]
# 4. Obtener las 10 palabras más frecuentes
frecuencia = FreqDist(palabras_limpias)
print(" 10 PALABRAS MÁS FRECUENTES ")
for palabra, cant in frecuencia.most_common(10):
    print(f"{palabra}: {cant}")
# 5. Mostrar el gráfico con frecuencia.plot(10)
frecuencia.plot(10, cumulative=False, title="10 Palabras Más Frecuentes en la Noticia")
plt.show()
```
