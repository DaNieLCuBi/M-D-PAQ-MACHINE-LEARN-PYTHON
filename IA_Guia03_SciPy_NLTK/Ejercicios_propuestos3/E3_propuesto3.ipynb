```python
import numpy as np
from scipy import linalg
# 1. Definir la matriz de coeficientes (A)
A = np.array([
    [1,  1,  1],
    [2, -1,  1],
    [1,  2, -1]])
# 2. Definir el vector de términos independientes (b)
b = np.array([6, 3, 2])
# 3. Resolver el sistema A * x = b
solucion = linalg.solve(A, b)
# 4. Extraer los valores de x, y, z
x, y, z = solucion
# 5. Mostrar los resultados
print("SOLUCIÓN DEL SISTEMA DE ECUACIONES ")
print(f"x = {x:.2f}") 
print(f"y = {y:.2f}") 
print(f"z = {z:.2f}\n")
# 6. Comprobación de las ecuaciones
print(" COMPROBACIÓN ")
print(f"Ecuación 1 (x + y + z = 6):      {x} + {y} + {z} = {x + y + z}")
print(f"Ecuación 2 (2x - y + z = 3):     2({x}) - ({y}) + ({z}) = {2*x - y + z}")
print(f"Ecuación 3 (x + 2y - z = 2):     {x} + 2({y}) - ({z}) = {x + 2*y - z}")
```
