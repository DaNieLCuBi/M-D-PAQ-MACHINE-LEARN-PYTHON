```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Entrenar un modelo base previo para ejecutar la validación
rng = np.random.default_rng(0)
horas_base = rng.uniform(1, 10, size=50).reshape(-1, 1)
nota_base = 6 + 1.3 * horas_base.ravel() + rng.normal(0, 1, size=50)

reg_interactivo = LinearRegression().fit(horas_base, nota_base)

try:
    horas_usuario = float(input("Ingrese la cantidad de horas de estudio (0 - 12): "))

    if 0 <= horas_usuario <= 12:
        nota_predicha = reg_interactivo.predict([[horas_usuario]])[0]
        print(f"Para {horas_usuario} horas de estudio, la nota estimada es: {nota_predicha:.2f}")
    else:
        print("Error: El valor ingresado debe estar comprendido entre 0 y 12.")
except ValueError:
    print("Error: Ingrese un valor numérico válido.")
```
