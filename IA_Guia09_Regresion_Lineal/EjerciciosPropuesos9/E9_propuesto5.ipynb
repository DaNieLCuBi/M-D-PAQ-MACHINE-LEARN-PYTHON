```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
# Datos simulados de estudiantes
rng = np.random.default_rng(0)
horas = rng.uniform(1, 10, size=50).reshape(-1, 1)
nota = 6 + 1.3 * horas.ravel() + rng.normal(0, 1, size=50)

X_train, X_test, y_train, y_test = train_test_split(
    horas, nota, test_size=0.2, random_state=0
)
# 1. Regresión Lineal
reg_modelo = LinearRegression().fit(X_train, y_train)
y_pred_reg = reg_modelo.predict(X_test)
mae_regresion = mean_absolute_error(y_test, y_pred_reg)
# 2. Modelo Baseline (Siempre predice el promedio del conjunto de entrenamiento)
promedio_train = y_train.mean()
y_pred_promedio = [promedio_train] * len(y_test)
mae_promedio = mean_absolute_error(y_test, y_pred_promedio)
# Mejora
mejora_absoluta = mae_promedio - mae_regresion
mejora_porcentual = (mejora_absoluta / mae_promedio) * 100

print("COMPARACIÓN DE MAE")
print(f"MAE del modelo basado en Promedio : {mae_promedio:.3f}")
print(f"MAE del modelo Regresión Lineal   : {mae_regresion:.3f}")
print(f"Mejora absoluta en error           : {mejora_absoluta:.3f} puntos")
print(f"Porcentaje de mejora del error     : {mejora_porcentual:.2f}%")
```
