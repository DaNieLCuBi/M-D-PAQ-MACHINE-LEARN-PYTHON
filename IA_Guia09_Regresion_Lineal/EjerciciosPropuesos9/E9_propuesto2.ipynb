```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Crear dataset de 30 registros
rng = np.random.default_rng(42)
temperatura = rng.uniform(18, 35, size=30).reshape(-1, 1)
# Ventas generadas con relación positiva y ruido
ventas = 50 + 12 * temperatura.ravel() + rng.normal(0, 15, size=30)

df_helados = pd.DataFrame(
    {"temperatura": temperatura.ravel(), "ventas": ventas}
)

# Entrenar regresión lineal simple
modelo_helados = LinearRegression()
modelo_helados.fit(temperatura, ventas)

# Predicción para 28 °C
temp_consulta = np.array([[28]])
venta_predicha = modelo_helados.predict(temp_consulta)[0]

print("PREDICCIÓN DE VENTAS DE HELADOS ")
print(f"Pendiente: {modelo_helados.coef_[0]:.2f}")
print(f"Intercepto: {modelo_helados.intercept_:.2f}")
print(f"Ventas estimadas para 28 °C: {venta_predicha:.2f} unidades")
```
