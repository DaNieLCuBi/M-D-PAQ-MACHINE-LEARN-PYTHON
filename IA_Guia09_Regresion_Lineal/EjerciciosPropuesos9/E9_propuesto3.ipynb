```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

# Datos simulados de viviendas
rng = np.random.default_rng(10)
n_casas = 100
m2 = rng.uniform(50, 200, n_casas)
cuartos = rng.integers(1, 5, n_casas)
distrito_premium = rng.choice([0, 1], size=n_casas, p=[0.7, 0.3])

# Precio depende de m2, cuartos y distrito_premium
precio = (
    50000
    + 1200 * m2
    + 15000 * cuartos
    + 45000 * distrito_premium
    + rng.normal(0, 10000, n_casas)
)

df_casas = pd.DataFrame(
    {
        "m2": m2,
        "cuartos": cuartos,
        "distrito_premium": distrito_premium,
        "precio": precio,
    }
)

# Modelo 1: Sin distrito_premium
X1 = df_casas[["m2", "cuartos"]]
y = df_casas["precio"]
X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y, test_size=0.2, random_state=42
)

reg1 = LinearRegression().fit(X1_train, y1_train)
r2_modelo1 = r2_score(y1_test, reg1.predict(X1_test))

# Modelo 2: Con distrito_premium
X2 = df_casas[["m2", "cuartos", "distrito_premium"]]
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y, test_size=0.2, random_state=42
)

reg2 = LinearRegression().fit(X2_train, y2_train)
r2_modelo2 = r2_score(y2_test, reg2.predict(X2_test))

print(" EVALUACIÓN DE INCLUSIÓN DE VARIABLE PREMIUM ")
print(f"R2 Sin 'distrito_premium': {r2_modelo1:.4f}")
print(f"R2 Con 'distrito_premium': {r2_modelo2:.4f}")
print(f"¿Mejora el R2?: {'Sí' if r2_modelo2 > r2_modelo1 else 'No'}")
```
