```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

# Datos simulados idénticos a la práctica base
rng = np.random.default_rng(0)
horas = rng.uniform(1, 10, size=50).reshape(-1, 1)
nota = 6 + 1.3 * horas.ravel() + rng.normal(0, 1, size=50)

# 1. Modelo con test_size = 0.2 (Original)
X_train20, X_test20, y_train20, y_test20 = train_test_split(
    horas, nota, test_size=0.2, random_state=0
)
reg20 = LinearRegression().fit(X_train20, y_train20)
y_pred20 = reg20.predict(X_test20)

# 2. Modelo con test_size = 0.3 (Modificado)
X_train30, X_test30, y_train30, y_test30 = train_test_split(
    horas, nota, test_size=0.3, random_state=0
)
reg30 = LinearRegression().fit(X_train30, y_train30)
y_pred30 = reg30.predict(X_test30)

print("COMPARACIÓN TEST_SIZE 0.2 VS 0.3")
print(
    f"test_size=0.2 -> Pendiente: {reg20.coef_[0]:.3f} | R2: {r2_score(y_test20, y_pred20):.3f}"
)
print(
    f"test_size=0.3 -> Pendiente: {reg30.coef_[0]:.3f} | R2: {r2_score(y_test30, y_pred30):.3f}"
)
```
