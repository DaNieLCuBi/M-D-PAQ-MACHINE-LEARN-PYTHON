```python
import numpy as np
import torch
import torch.nn as nn
# Generación de los datos base (horas vs nota)
rng = np.random.default_rng(0)
horas = rng.uniform(1, 10, size=50).reshape(-1, 1).astype(np.float32)
nota = (6.0 + 1.3 * horas + rng.normal(0, 1, size=(50, 1))).astype(np.float32)

X = torch.from_numpy(horas)
y = torch.from_numpy(nota)

def entrenar(lr_val, epocas=150):
    torch.manual_seed(42)  # Misma inicialización para todos
    modelo = nn.Linear(1, 1)
    criterio = nn.MSELoss()
    optimizador = torch.optim.Adam(modelo.parameters(), lr=lr_val)

    perdidas = []
    for epoca in range(1, epocas + 1):
        modelo.train()
        pred = modelo(X)
        loss = criterio(pred, y)

        optimizador.zero_grad()
        loss.backward()
        optimizador.step()
        perdidas.append(loss.item())
    peso = modelo.weight.item()
    sesgo = modelo.bias.item()
    return perdidas[0], perdidas[-1], peso, sesgo

tasas = [0.001, 0.01, 0.05]
print(
    f"{'lr':<8} | {'Pérdida inicial':<16} | {'Pérdida final':<16} | {'Peso (w)':<10} | {'Sesgo (b)':<10}"
) print("-" * 70)
for lr in tasas:
    p_ini, p_fin, w, b = entrenar(lr)
    print(f"{lr:<8} | {p_ini:<16.4f} | {p_fin:<16.4f} | {w:<10.2f} | {b:<10.2f}")
```
