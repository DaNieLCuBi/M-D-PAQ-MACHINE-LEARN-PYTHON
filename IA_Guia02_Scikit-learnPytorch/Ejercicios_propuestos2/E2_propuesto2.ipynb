```python
from sklearn.datasets import load_wine
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
# 1. Cargar el dataset de vinos
X, y = load_wine(return_X_y=True)
# 2. Definición del diccionario de modelos (incluyendo RandomForestClassifier)
modelos = {
    "Regresión logística": make_pipeline(StandardScaler(), LogisticRegression(max_iter=200, random_state=42)),
    "Árbol de decisión": DecisionTreeClassifier(max_depth=3, random_state=42),
    "KNN (k=5)": make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5)),
    "Random Forest": RandomForestClassifier(random_state=42),
}
# 3. Evaluación con validación cruzada (cv=5)
print("COMPARACIÓN EN DATASET WINE (cv=5)")
print(f"{'Modelo':<24} | {'Promedio Exactitud':<20} | {'Desviación':<15}")
print("-" * 65)
mejores_puntajes = {}
for nombre, modelo in modelos.items():
    puntajes = cross_val_score(modelo, X, y, cv=5)
    media = puntajes.mean()
    desv = puntajes.std()
    mejores_puntajes[nombre] = media
    print(f"{nombre:<24} | {media:.4f} ({media*100:.2f}%)       | +/- {desv:.4f}")
# 4. Identificar el mejor
ganador = max(mejores_puntajes, key=mejores_puntajes.get)
print("-" * 65)
print(f"El modelo con mejor rendimiento promedio es: {ganador} ({mejores_puntajes[ganador]*100:.2f}%)")
```
