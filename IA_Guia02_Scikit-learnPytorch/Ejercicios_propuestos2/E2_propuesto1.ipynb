```python
from sklearn.datasets import load_iris
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# Carga de datos
X, y = load_iris(return_X_y=True)

valores_k = [1, 3, 7, 15]

print(f"{'n_neighbors':<12} | {'Promedio Exactitud (Accuracy)':<30} | {'Desviación Estándar':<20}")
print("-" * 68)

resultados = {}
for k in valores_k:
    modelo_knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=k))
    puntajes = cross_val_score(modelo_knn, X, y, cv=5)
    media = puntajes.mean()
    desv = puntajes.std()
    resultados[k] = (media, desv)
    print(f"k = {k:<9} | {media:.4f} ({media*100:.2f}%)               | +/- {desv:.4f}")
```
