```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

def evaluar_regresion(ruido_std):
    rng = np.random.default_rng(0)
    horas = rng.uniform(1, 10, size=50).reshape(-1, 1)
    # Simulación de la nota con el ruido indicado
    ruido = rng.normal(0, ruido_std, size=50)
    nota = 6 + 1.3 * horas.ravel() + ruido
    
    X_train, X_test, y_train, y_test = train_test_split(
        horas, nota, test_size=0.2, random_state=0
    )
    reg = LinearRegression()
    reg.fit(X_train, y_train)

    y_pred = reg.predict(X_test)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)

    return mae, r2, reg.coef_[0], reg.intercept_

# Evaluación de ambos escenarios
mae_orig, r2_orig, coef_orig, inter_orig = evaluar_regresion(ruido_std=1)
mae_ruido, r2_ruido, coef_ruido, inter_ruido = evaluar_regresion(ruido_std=3)

print("COMPARACIÓN DE RESULTADOS")
print(f"{'Métrica':<25} | {'Original (std=1)':<18} | {'Con más ruido (std=3)':<20}")
print("-" * 70)
print(f"{'Error Absoluto Medio (MAE)':<25} | {mae_orig:<18.3f} | {mae_ruido:<20.3f}")
print(f"{'Coeficiente R²':<25} | {r2_orig:<18.3f} | {r2_ruido:<20.3f}")
print(f"{'Pendiente estimada':<25} | {coef_orig:<18.3f} | {coef_ruido:<20.3f}")
print(f"{'Intercepto estimado':<25} | {inter_orig:<18.3f} | {inter_ruido:<20.3f}")
```
